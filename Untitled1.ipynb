```python
import matplotlib.pyplot as plt
import csv

x = []
y = []

with open('pgmt.csv', 'r') as csvfile:
    lines = csv.reader(csvfile, delimiter=',')
    # Skip the header row ('Date', 'Temperature')
    next(lines, None)

    for row in lines:
        x.append(row[0])
        y.append(int(row[1]))

print(x, y)

plt.plot(x, y, color='r', linestyle='-.',
         marker='*', ms=20, mfc='b', mec='g',
         label='Weather Data')

plt.xticks(rotation=25)
plt.xlabel('Dates')
plt.ylabel('Temperature')
plt.title('Weather Report', fontsize=20)
plt.grid()
plt.legend()
plt.show()
```
